```python
import pandas as pd
df = pd.read_csv('data/raw/train_transaction.csv')
```

```python
print("Loading data...")
df = df.sort_values("TransactionDT").reset_index(drop=True)
print("Total rows:", len(df))
n = len(df)
train_end = int(n * 0.70)
validation_end = int(n * 0.85)
train = df.iloc[:train_end].copy()
validation = df.iloc[train_end:validation_end].copy()
test = df.iloc[validation_end:].copy()
print("\n========== SPLIT SIZES ==========")
print("Train:", train.shape)
print("Validation:", validation.shape)
print("Test:", test.shape)
print(
    "Train:",
    round(train["isFraud"].mean() * 100, 3),
    "%"
)
print(
    "Validation:",
    round(validation["isFraud"].mean() * 100, 3),
    "%"
)
print(
    "Test:",
    round(test["isFraud"].mean() * 100, 3),
    "%"
)
print("\n========== TIME RANGES ==========")
print(
    "Train:",
    train["TransactionDT"].min(),
    "→",
    train["TransactionDT"].max()
)
print(
    "Validation:",
    validation["TransactionDT"].min(),
    "→",
    validation["TransactionDT"].max()
)
print(
    "Test:",
    test["TransactionDT"].min(),
    "→",
    test["TransactionDT"].max()
)
```

```python
print("\n========== CATEGORICAL COLUMNS ==========")
categorical_columns = train.select_dtypes(
    include=["object"]
).columns.tolist()
print("Number:", len(categorical_columns))
for column in categorical_columns:
    print(
        f"{column:20}"
        f"unique={train[column].nunique(dropna=False):5}"
    )
print("\n========== NUMERICAL COLUMNS ==========")
numeric_columns = train.select_dtypes(
    include=["number"]
).columns.tolist()
print("Number:", len(numeric_columns))
```
